# Tabular Q-learning for a Double Pendulum

This notebook runs the commented implementation from `qlearning_double_pendulum.py`. The continuous state is discretized into two angle bins and two angular-velocity bins. Evaluation uses the greedy policy after training.

```python
%matplotlib inline
import numpy as np
import qlearning_double_pendulum as ql

# Change this value for a quick test or a longer training run.
EPISODES = 30000
print(f"State table: {ql.ANGLE_BINS} x {ql.ANGLE_BINS} x {ql.VELOCITY_BINS} x {ql.VELOCITY_BINS}")
print(f"Actions: {len(ql.ACTION_TORQUES)} | Horizon: {ql.HORIZON}")
```

## Training

The Q-learning update uses the discretized next state and the maximum Q-value over the nine joint-torque combinations.

```python
# Train the table with epsilon-greedy exploration.
q_table, returns = ql.train(EPISODES)
print(f"Final mean return (last 100): {np.mean(returns[-100:]):.3f}")
```

## Training reward

```python
# Plot episode returns and their moving average.
ql.plot_rewards(returns)
```

## Greedy evaluation and visualization

Exploration is disabled during evaluation: the action with the largest Q-value in the current discretized state is selected.

```python
# Animate three greedy evaluation episodes after training.
animation = ql.visualize(q_table, episodes=3)
```
